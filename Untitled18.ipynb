```python
# 1 Missing values
df["flag_missing_values"] = df[
    ["companynameofficial","timevalue","REVENUE"]
].isnull().any(axis=1)
```

```python
# 2 Invalid revenue
df["flag_invalid_revenue"] = df["REVENUE"] <= 0
```

```python
# 3 Revenue outlier
df = df.sort_values(["companynameofficial","timevalue"])

df["revenue_growth"] = (
    df.groupby("companynameofficial")["REVENUE"]
    .pct_change()
)

df["flag_revenue_outlier"] = df["revenue_growth"].abs() > 5
```

```python
# 4 Currency inconsistency
df["flag_currency_inconsistency"] = (
    df.groupby("companynameofficial")["unit_REVENUE"]
    .transform(lambda x: x.nunique() > 1)
)
```

```python
import os

# 5 AI / LLM style check
def company_country_check(company, country):

    known = {
        "BMW":"Germany",
        "Toyota":"Japan",
        "Apple":"United States"
    }

    for k in known:
        if k.lower() in str(company).lower():
            return known[k] != country

    return False


df["flag_country_anomaly"] = df.apply(
    lambda row: company_country_check(
        row["companynameofficial"],
        row["geonameen"]
    ),
    axis=1
)


df["flag_any_issue"] = df[[
"flag_missing_values",
"flag_invalid_revenue",
"flag_revenue_outlier",
"flag_currency_inconsistency",
"flag_country_anomaly"
]].any(axis=1)

os.makedirs('../output', exist_ok=True)
df.to_excel("../output/quality_checked_output.xlsx", index=False)

print("Finished. Output created.")
```
